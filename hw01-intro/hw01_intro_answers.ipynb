This notebook contains the answers for the intro homework questions using the dataset `car_fuel_efficiency_2026.csv`.

```python
import pandas as pd

df = pd.read_csv('../car_fuel_efficiency_2026.csv')
df.head()
```

## 4. Maximum fuel efficiency among cars from Asia

```python
df.loc[df['origin'] == 'Asia', 'fuel_efficiency_mpg'].max()
```

## 5. Median value of horsepower

```python
median_before = df['horsepower'].median()
median_before
```

## 6. Most frequent value of horsepower

```python
most_frequent = df['horsepower'].mode()[0]
most_frequent
```

## 7. Fill missing horsepower values with the most frequent value and recompute median

```python
df['horsepower'] = df['horsepower'].fillna(most_frequent)
df['horsepower'].median()
```

## 8. Matrix multiplication exercise

```python
import numpy as np

asia = df[df['origin'] == 'Asia']
X_df = asia[['vehicle_weight', 'model_year']].head(7)
X = X_df.to_numpy()
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200], dtype=float)
w = XTX_inv.dot(X.T).dot(y)
print(w)
print(w.sum())
```
